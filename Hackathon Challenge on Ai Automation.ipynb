### Average Salary by Industry


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\kashr\OneDrive\Documents\ai_job_market_insights.csv")
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="Industry",
    y="Salary_USD"
)

plt.title("Average Salary by Industry")
plt.xticks(rotation=45)
plt.show()
```

```python
automation_risk = (
    df.groupby("Industry")["Automation_Risk"]
      .value_counts(normalize=True)
      .unstack()
)

automation_risk["High"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("High Automation Risk by Industry")
plt.xlabel("Industry")
plt.ylabel("Proportion of Jobs")
plt.xticks(rotation=45)
plt.show()
```

```python
skill_counts = df["Required_Skills"].value_counts()

skill_counts.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Most Frequently Required Skills")
plt.xlabel("Number of Jobs")
plt.ylabel("Required Skill")
plt.show()
```

```python
growth_by_ai = (
    df.groupby("AI_Adoption_Level")["Job_Growth_Projection"]
      .value_counts(normalize=True)
      .unstack()
)

growth_by_ai.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Job Growth Projection by AI Adoption Level")
plt.xlabel("AI Adoption Level")
plt.ylabel("Proportion of Jobs")
plt.xticks(rotation=0)
plt.legend(title="Job Growth Projection")
plt.show()
```

```python
risk_by_ai = (
    df.groupby("AI_Adoption_Level")["Automation_Risk"]
      .value_counts(normalize=True)
      .unstack()
)

risk_by_ai.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Automation Risk by AI Adoption Level")
plt.xlabel("AI Adoption Level")
plt.ylabel("Proportion of Jobs")
plt.xticks(rotation=0)
plt.legend(title="Automation Risk")
plt.show()
```

# Key findings

Base on my findings from the 500 raw dataset provided

Jobs in Transportation had the highest proportion of high automation risk at 48.7%.
High AI adoption jobs had the highest proportion of projected job decline at 37.4%.
Low AI adoption jobs had the highest proportions of projected growth and stability, both at 35.1%.
Low AI adoption jobs had the highest average salary at approximately $93,354, while high AI adoption jobs averaged approximately $87,583.
JavaScript had the highest average salary among the required-skill categories at approximately $98,944.
Python and Project Management were among the most frequently required skills, each appearing in 60 jobs.
The relationship between AI adoption and automation risk was not simply linear; medium AI adoption had the highest proportion of high-risk jobs at 38.0%.
The dataset suggests that workers may need to consider both current job demand and emerging skills when preparing for changes in the labour market.
